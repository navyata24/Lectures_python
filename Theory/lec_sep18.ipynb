```python
def addition(a,b):
    add=a+b
    return add
def substraction(a,b):
    sub=a-b
    return sub
def multiplication(a,b):
    multiply=a*b
    return multiply
def division(a,b):
    if b == 0:
        return "Error: Cannot divide by zero!"
    div=a/b
    return div
a=int(input("Enter a number: "))
b=int(input("Enter a number: "))

# Call the functions and store their results in new variables
result_add = addition(a, b)
result_sub = substraction(a, b)
result_multiply = multiplication(a, b)
result_div = division(a, b)

# Print the results
print(f"Addition: {result_add}")
print(f"Subtraction: {result_sub}")
print(f"Multiplication: {result_multiply}")
print(f"Division: {result_div}")
```

```python
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    if b == 0:
        return "Error: Cannot divide by zero!"
    return a / b


a = int(input("Enter a number: "))
b = int(input("Enter a number: "))

print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter your choice: "))

if choice == 1:
    print("Addition:", addition(a, b))

elif choice == 2:
    print("Subtraction:", subtraction(a, b))

elif choice == 3:
    print("Multiplication:", multiplication(a, b))

elif choice == 4:
    print("Division:", division(a, b))

else:
    print("Invalid choice")
```
