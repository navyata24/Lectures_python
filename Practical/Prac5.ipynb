```python
marks = int(input("Enter marks: "))
for marks in range(0, 101):
  percentage=(marks/100)*100
if percentage>=90:
    grade="A"
elif percentage>=75:
   grade="B"
elif percentage>=50:
   grade="C"
else:
    grade="F"
print("Grade:",grade)
```
