Q1. A dictionary maps 4 Indian states to their capitals. Write a program that asks the user for the capital of each state,
compares the answers case-insensitively after removing extra spaces, keeps the score, and finally prints the score, the
percentage and a message ("Excellent" for 75% or more, "Good" for 50-74%, otherwise "Keep practising").

```python
capitals={
    "Maharashtra":"Mumbai",
    "Rajasthan":"Jaipur",
    "Uttarakhand":"Dehradun",
    "Punjab":"Chandigarh"
}
score = 0
for state,capital in capitals.items():
    answer = input("Enter the capital of " + state + ": ")
    if answer.strip().lower() == capital.lower():
        print("Correct!")
        score += 1
    else:
        print("Incorrect!")
percentage = (score / len(capitals)) * 100
print("Score:", score, "out of", len(capitals))
print("Percentage:", percentage, "%")
if percentage >= 75:
    print("Excellent")
elif percentage >= 50:
    print("Good")
else:
    print("Keep practising")
```

Q2. Write a program that accepts the number of rows n and prints (a) a right-angled number triangle in which row i
contains the numbers 1 to i, and (b) an inverted triangle of stars in which row 1 has n stars and the last row has 1 star.

```python
n = int(input("Enter the number of rows: "))

#(a)Right-angled number triangle
print("Number Triangle:")
for i in range(1,n + 1):
    for j in range(1,i + 1):
        print(j,end=" ")
    print()

#(b)Inverted star triangle
print("Inverted Star Triangle:")
for i in range(n,0,-1):
    print("* " * i)
```
