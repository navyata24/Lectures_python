2. Let word="Data". Print word*3 , and add a comment explaining what the * operator does when used with a string and a number.

```python
#Q2

word="Data"
print(word*3)

# When '*' operator is used with a string and a number , it prints the string as many times as the number entered.
# Here the word "Data" is printed thrice with no space in between as it is multiplied by 3('*' is used for multiplication in arithmetic operations)
```

3. Let lst=[1,2,3] and lst2=lst(NOT a copy) . Do lst2.append(4) and print both lst and lst2 - expalin in a comment why BOTH lists changed.

```python
#Q3

lst=[1,2,3]
lst2=lst # Both variables refer to the same list , so changes through lst2 affect lst.Here lst2=lst does not copy the list but instead both variables point to the same list.
lst2.append(4)
print(lst)
print(lst2)
```
